```python
ft_model = fasttext.load_model("cc.id.300.bin")
def get_synonym_fasttext(word, model, topn=5):
    try:
        neighbors = model.get_nearest_neighbors(word, k=topn)
        return [w for _, w in neighbors]
    except:
        return []
```

```python
def synonym_replacement_ft(text, model):
    if not isinstance(text, str):
        return text

    words = text.lower().split()
    new_words = words.copy()

    for i, word in enumerate(words):
        synonyms = get_synonym_fasttext(word, model)
        if synonyms:
            new_words[i] = random.choice(synonyms)

    return " ".join(new_words)
```

```python
# Menukar posisi kata secara acak
def word_swap(text):
    words = text.split()
    if len(words) < 2:
        return text

    i, j = random.sample(range(len(words)), 2)
    words[i], words[j] = words[j], words[i]
    return " ".join(words)
```

```python
# Menghapus baris data yang kosong
df = df.dropna(subset=['Gejala', 'Diagnosa', 'Penanganan'])
```

```python
df['Gejala'] = df['Gejala'].astype(str)
df['Diagnosa'] = df['Diagnosa'].astype(str)
df['Penanganan'] = df['Penanganan'].astype(str)
```

```python
def augment_gejala(text):
    augmented = set()
    augmented.add(text.lower())
    augmented.add(text.replace(",", " "))
    augmented.add(" ".join(text.split()))
    return list(augmented)
```

```python
def augment_gejala(text):
    words = text.split()
    augmented = set()
    augmented.add(text)

    if len(words) > 3:
        idx = random.randint(0, len(words)-1)
        augmented.add(" ".join(words[:idx] + words[idx+1:]))

    return list(augmented)
```

```python
import random
df = df.dropna(subset=['Gejala'])

augmented_rows = []

for row in df.itertuples(index=False):
    for g in augment_gejala(row.Gejala):
        augmented_rows.append(
            {
                "Gejala": g,
                "Diagnosa": row.Diagnosa,
                "Penanganan": row.Penanganan,
            }
        )

df_aug = pd.DataFrame(augmented_rows)
```

PRE PROCESSING DATA

```python
stemmer = StemmerFactory().create_stemmer()
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())
```

```python
def preprocess_text(text):
    # 1. lowercase
    text = text.lower()

    # 2. hapus angka & simbol
    text = re.sub(r'[^a-z\s]', ' ', text)

    # 3. tokenisasi
    tokens = text.split()

    # 4. stopword removal
    tokens = [t for t in tokens if t not in stopwords]

    # 5. stemming
    tokens = [stemmer.stem(t) for t in tokens]

    return " ".join(tokens)
```

```python
df_aug['Gejala_Preprocessed'] = df_aug['Gejala'].apply(preprocess_text)
```

TF IDF

```python
tfidf = TfidfVectorizer(
    max_df=0.95,
    min_df=2
)

tfidf_matrix = tfidf.fit_transform(df_aug['Gejala_Preprocessed'])
```

LSI


```python
n_components = 100
svd = TruncatedSVD(n_components=n_components, random_state=42)

lsi_matrix = svd.fit_transform(tfidf_matrix)
```

COSINE SIMILARITY

```python
def cari_diagnosa_interaktif(query, top_k=5):
    # Preprocessing query
    query_clean = preprocess_text(query)
    query_tfidf = tfidf.transform([query_clean])
    query_lsi = svd.transform(query_tfidf)

    # Cosine similarity
    similarity = cosine_similarity(query_lsi, lsi_matrix)[0]

    df_temp = df_aug.copy()
    df_temp['Similarity'] = similarity

    # Ambil gejala terbaik per diagnosa
    idx = df_temp.groupby('Diagnosa')['Similarity'].idxmax()
    ranking = df_temp.loc[idx][
        ['Diagnosa', 'Penanganan', 'Gejala', 'Similarity']
    ]

    # Threshold
    THRESHOLD = 0.25
    ranking = ranking[ranking['Similarity'] >= THRESHOLD]

    # Sorting & ranking
    ranking = ranking.sort_values(
        by='Similarity', ascending=False
    ).reset_index(drop=True)

    ranking['Rank'] = ranking.index + 1

    # Top-K
    ranking = ranking.head(top_k)

    return ranking[['Rank', 'Gejala', 'Diagnosa', 'Penanganan', 'Similarity']]
```

```python
query = input("Masukkan gejala sapi: ")
hasil = cari_diagnosa_interaktif(query)
hasil
```
